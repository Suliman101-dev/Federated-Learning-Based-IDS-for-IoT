```python
"""
server.py
---------
Flower federated learning server for the BoT-IoT FT-Transformer.

Usage:
    python server.py --num_rounds 10 \
                     --num_clients 5 \
                     --data_dir ./client_data \
                     --min_fit_clients 3

The server:
  - Runs FedAvg aggregation
  - Evaluates the global model centrally after each round
    (using a held-out test split from the full data, saved during sharding)
  - Plots and saves training curves at the end
"""

import argparse
import json
import os
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from collections import OrderedDict
from typing import List, Tuple, Optional, Dict
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

import flwr as fl
from flwr.common import (
    Parameters, Scalar, NDArrays,
    ndarrays_to_parameters, parameters_to_ndarrays
)
from flwr.server.strategy import FedAvg
from flwr.server.client_proxy import ClientProxy
from flwr.common import FitRes, EvaluateRes


# ─────────────────────────────────────────────────────────────
# FT-Transformer (must match client)
# ─────────────────────────────────────────────────────────────

class FTTransformer(nn.Module):
    def __init__(self, n_features, d_model=128, n_heads=8,
                 n_layers=6, d_ffn=512, n_classes=5, dropout=0.1):
        super().__init__()
        self.feature_embeddings = nn.ModuleList([
            nn.Linear(1, d_model) for _ in range(n_features)
        ])
        self.cls_token = nn.Parameter(torch.randn(1, 1, d_model))
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=n_heads,
            dim_feedforward=d_ffn, dropout=dropout,
            batch_first=True
        )
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)
        self.head = nn.Sequential(
            nn.LayerNorm(d_model), nn.ReLU(),
            nn.Dropout(dropout), nn.Linear(d_model, n_classes)
        )

    def forward(self, x):
        batch_size = x.size(0)
        embeddings = [self.feature_embeddings[i](x[:, i:i+1])
                      for i in range(x.size(1))]
        embeddings = torch.stack(embeddings, dim=1)
        cls = self.cls_token.expand(batch_size, -1, -1)
        embeddings = torch.cat([cls, embeddings], dim=1)
        embeddings = self.transformer(embeddings)
        return self.head(embeddings[:, 0])


# ─────────────────────────────────────────────────────────────
# Helpers: parameters ↔ model
# ─────────────────────────────────────────────────────────────

def get_parameters(model: nn.Module) -> NDArrays:
    return [val.cpu().numpy() for _, val in model.state_dict().items()]


def set_parameters(model: nn.Module, parameters: NDArrays) -> None:
    params_dict = zip(model.state_dict().keys(), parameters)
    state_dict = OrderedDict({k: torch.tensor(v) for k, v in params_dict})
    model.load_state_dict(state_dict, strict=True)


# ─────────────────────────────────────────────────────────────
# Central evaluation on server-side test set
# ─────────────────────────────────────────────────────────────

from sklearn.metrics import f1_score, precision_score, recall_score, roc_auc_score, confusion_matrix

def build_evaluate_fn(model, test_loader, device, output_dir):
    history = []  # ← add this

    def evaluate(server_round, parameters, config):
        set_parameters(model, parameters)
        model.eval()
        model.to(device)

        criterion = nn.CrossEntropyLoss()
        total_loss, total = 0.0, 0
        all_preds, all_labels, all_probs = [], [], []

        with torch.no_grad():
            for X_batch, y_batch in test_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)
                out = model(X_batch)
                total_loss += criterion(out, y_batch).item() * len(y_batch)
                total += len(y_batch)
                all_preds.extend(out.argmax(1).cpu().numpy())
                all_labels.extend(y_batch.cpu().numpy())
                all_probs.extend(F.softmax(out, dim=1).cpu().numpy().tolist())  # ← add

        loss     = total_loss / total
        acc      = np.mean(np.array(all_preds) == np.array(all_labels))
        f1       = f1_score(all_labels, all_preds, average='weighted', zero_division=0)
        precision = precision_score(all_labels, all_preds, average='weighted', zero_division=0)
        recall   = recall_score(all_labels, all_preds, average='weighted', zero_division=0)
        cm       = confusion_matrix(all_labels, all_preds).tolist()
        try:
            auc = roc_auc_score(all_labels, all_probs, multi_class='ovr', average='weighted')
            print('auc caluclated')
        except:
            auc = 0.0
            print('auc is not caluclated')

        print(f"\n[Server] Round {server_round} → Loss: {loss:.4f} | Acc: {acc:.4f} | "
              f"F1: {f1:.4f} | AUC: {auc:.4f}\n")

        # Save to JSON
        history.append({"round": server_round, "loss": loss, "accuracy": acc,
                         "f1": f1, "precision": precision, "recall": recall,
                         "auc": auc, "confusion_matrix": cm})
        with open(os.path.join(output_dir, 'server_metrics.json'), 'w') as f:
            json.dump(history, f, indent=2)

        return loss, {"central_accuracy": acc, "f1": f1, "precision": precision,
                      "recall": recall, "auc": auc}

    return evaluate


# ─────────────────────────────────────────────────────────────
# Custom FedAvg strategy with metric tracking
# ─────────────────────────────────────────────────────────────

class TrackingFedAvg(FedAvg):
    """FedAvg with round-level metric tracking for plotting."""

    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.round_metrics = {
            'round': [],
            'distributed_train_loss': [],
            'distributed_train_f1': [],
            'distributed_eval_loss': [],
            'distributed_eval_accuracy': [],
            'central_loss': [],
            'central_accuracy': []
        }

    def aggregate_fit(
        self,
        server_round: int,
        results: List[Tuple[ClientProxy, FitRes]],
        failures
    ):
        # Aggregate weights
        aggregated = super().aggregate_fit(server_round, results, failures)

        # Collect distributed training metrics
        if results:
            losses = [r.metrics.get('train_loss', 0.0) * r.num_examples
                      for _, r in results]
            f1s    = [r.metrics.get('train_f1', 0.0) * r.num_examples
                      for _, r in results]
            total  = sum(r.num_examples for _, r in results)
            avg_loss = sum(losses) / total if total > 0 else 0.0
            avg_f1   = sum(f1s)    / total if total > 0 else 0.0

            self.round_metrics['distributed_train_loss'].append(avg_loss)
            self.round_metrics['distributed_train_f1'].append(avg_f1)
            print(f"[Server] Round {server_round} | "
                  f"Agg Train Loss: {avg_loss:.4f} | Agg Train F1: {avg_f1:.4f}")

        return aggregated

    def aggregate_evaluate(
        self,
        server_round: int,
        results: List[Tuple[ClientProxy, EvaluateRes]],
        failures
    ):
        aggregated = super().aggregate_evaluate(server_round, results, failures)

        if results:
            losses = [r.loss * r.num_examples for _, r in results]
            accs   = [r.metrics.get('accuracy', 0.0) * r.num_examples
                      for _, r in results]
            total  = sum(r.num_examples for _, r in results)
            avg_loss = sum(losses) / total if total > 0 else 0.0
            avg_acc  = sum(accs)   / total if total > 0 else 0.0

            self.round_metrics['distributed_eval_loss'].append(avg_loss)
            self.round_metrics['distributed_eval_accuracy'].append(avg_acc)
            self.round_metrics['round'].append(server_round)
            print(f"[Server] Round {server_round} | "
                  f"Agg Eval Loss: {avg_loss:.4f} | Agg Eval Acc: {avg_acc:.4f}")

        return aggregated


# ─────────────────────────────────────────────────────────────
# Post-training: plot and save
# ─────────────────────────────────────────────────────────────

def plot_and_save(strategy: TrackingFedAvg, model, output_dir: str):
    os.makedirs(output_dir, exist_ok=True)

    metrics = strategy.round_metrics
    rounds  = metrics.get('round', list(range(1, len(metrics['distributed_eval_accuracy']) + 1)))

    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.suptitle('Federated Learning – Global Model Results', fontsize=16, fontweight='bold')

    # 1. Distributed train loss
    if metrics['distributed_train_loss']:
        ax = axes[0, 0]
        ax.plot(metrics['distributed_train_loss'], 'b-o', linewidth=2, markersize=5)
        ax.set_title('Aggregated Client Train Loss', fontweight='bold')
        ax.set_xlabel('Round'); ax.set_ylabel('Loss')
        ax.grid(alpha=0.3)

    # 2. Distributed eval accuracy
    ax = axes[0, 1]
    ax.plot(rounds, metrics['distributed_eval_accuracy'], 'g-o', linewidth=2, markersize=5)
    ax.set_title('Aggregated Client Eval Accuracy', fontweight='bold')
    ax.set_xlabel('Round'); ax.set_ylabel('Accuracy')
    ax.set_ylim(0, 1); ax.grid(alpha=0.3)

    # 3. Distributed train F1
    if metrics['distributed_train_f1']:
        ax = axes[1, 0]
        ax.plot(metrics['distributed_train_f1'], 'r-o', linewidth=2, markersize=5)
        ax.set_title('Aggregated Client Train F1', fontweight='bold')
        ax.set_xlabel('Round'); ax.set_ylabel('F1-Score')
        ax.set_ylim(0, 1); ax.grid(alpha=0.3)

    # 4. Distributed eval loss
    ax = axes[1, 1]
    ax.plot(rounds, metrics['distributed_eval_loss'], 'm-o', linewidth=2, markersize=5)
    ax.set_title('Aggregated Client Eval Loss', fontweight='bold')
    ax.set_xlabel('Round'); ax.set_ylabel('Loss')
    ax.grid(alpha=0.3)

    plt.tight_layout()
    plot_path = os.path.join(output_dir, 'fl_training_curves.png')
    plt.savefig(plot_path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"Training curves saved → {plot_path}")

    # Save global model
    model_path = os.path.join(output_dir, 'global_model.pth')
    torch.save(model.state_dict(), model_path)
    print(f"Global model saved → {model_path}")

    # Save metrics JSON
    metrics_path = os.path.join(output_dir, 'fl_metrics.json')
    with open(metrics_path, 'w') as f:
        json.dump({k: [float(v) for v in vals]
                   for k, vals in metrics.items()}, f, indent=2)
    print(f"Metrics JSON saved → {metrics_path}")


# ─────────────────────────────────────────────────────────────
# Main
# ─────────────────────────────────────────────────────────────
# ─────────────────────────────────────────────────────────────
# Run in Notebook
# ─────────────────────────────────────────────────────────────

# Set parameters directly
num_rounds = 5
num_clients = 5
min_fit_clients = 5
data_dir = '/content/client_data'
output_dir = './fl_outputs'
server_address = '0.0.0.0:9090'
fraction_fit = 1.0
fraction_eval = 1.0

# Load metadata
meta_path = os.path.join(data_dir, 'meta.json')
with open(meta_path) as f:
    meta = json.load(f)

os.makedirs(output_dir, exist_ok=True)

n_features = meta['n_features']
n_classes = meta['n_classes']
print(f"\n{'='*60}")
print(f"  FL Server – BoT-IoT FT-Transformer")
print(f"{'='*60}")
print(f"  Features   : {n_features}")
print(f"  Classes    : {n_classes} {meta['classes']}")
print(f"  Rounds     : {num_rounds}")
print(f"  Clients    : {num_clients}")
print(f"  Min fit    : {min_fit_clients}")
print(f"{'='*60}\n")

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Build global model
global_model = FTTransformer(
    n_features=n_features, n_classes=n_classes
).to(device)

# Load test set if available
test_X_path = os.path.join(data_dir, 'test_X.npy')
test_y_path = os.path.join(data_dir, 'test_y.npy')
evaluate_fn = None
if os.path.exists(test_X_path) and os.path.exists(test_y_path):
    X_test = np.load(test_X_path)
    y_test = np.load(test_y_path)
    test_dataset = TensorDataset(
        torch.FloatTensor(X_test), torch.LongTensor(y_test)
    )
    test_loader = DataLoader(test_dataset, batch_size=4096, shuffle=False)
    #evaluate_fn = build_evaluate_fn(global_model, test_loader, device)
    evaluate_fn = build_evaluate_fn(global_model, test_loader, device, output_dir)  
    print(f"Central test set loaded: {len(X_test):,} samples\n")

# Initial parameters
initial_params = ndarrays_to_parameters(get_parameters(global_model))

# Strategy
strategy = TrackingFedAvg(
    fraction_fit=fraction_fit,
    fraction_evaluate=fraction_eval,
    min_fit_clients=min_fit_clients,
    min_evaluate_clients=min_fit_clients,
    min_available_clients=min_fit_clients,
    evaluate_fn=evaluate_fn,
    initial_parameters=initial_params,
    fit_metrics_aggregation_fn=lambda metrics: {},
    evaluate_metrics_aggregation_fn=lambda metrics: {
        "accuracy": np.average(
            [m["accuracy"] for _, m in metrics],
            weights=[n for n, _ in metrics]
        )
    }
)

# Start server
print(f"Starting server on {server_address} ...\n")
fl.server.start_server(
    server_address=server_address,
    config=fl.server.ServerConfig(num_rounds=num_rounds),
    strategy=strategy
)

# After training: save outputs
print("\nTraining complete. Saving outputs ...")
plot_and_save(strategy, global_model, output_dir)
```
